# Robustesse image (simulation)

Mesure de l'effet d'un bruit additif synthétique sur un score de détection. Les images et le détecteur sont simulés.

```python
from pathlib import Path
import sys

ROOT = next(parent for parent in [Path.cwd(), *Path.cwd().parents] if (parent / "src").is_dir())
sys.path.insert(0, str(ROOT))

import matplotlib.pyplot as plt
import numpy as np
from src.metrics import roc_metrics
```

```python
rng = np.random.default_rng(12)
y_true = np.tile([0, 1], 250)
base_signal = y_true.astype(float) + rng.normal(0, 0.6, y_true.size)
noise_levels = [0.0, 0.25, 0.5, 1.0, 1.5]
aucs = []
for noise in noise_levels:
    perturbed_score = base_signal + rng.normal(0, noise, y_true.size)
    aucs.append(roc_metrics(y_true, perturbed_score)["auc"])
list(zip(noise_levels, aucs))
```

```python
plt.plot(noise_levels, aucs, marker="o")
plt.xlabel("Écart-type du bruit ajouté")
plt.ylabel("AUC")
plt.title("Robustesse du score jouet")
plt.ylim(0.45, 1.0)
plt.grid(alpha=0.25)
plt.show()
```
