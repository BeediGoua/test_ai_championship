# KGW metrics (exemple jouet)

Démonstration des métriques ROC sur des scores synthétiques. Ce notebook ne contient pas d'implémentation KGW.

```python
from pathlib import Path
import sys

ROOT = next(parent for parent in [Path.cwd(), *Path.cwd().parents] if (parent / "src").is_dir())
sys.path.insert(0, str(ROOT))

import matplotlib.pyplot as plt
import numpy as np
from src.metrics import roc_metrics
```

```python
rng = np.random.default_rng(42)
y_true = np.r_[np.zeros(200, dtype=int), np.ones(200, dtype=int)]
y_score = np.r_[rng.normal(0.35, 0.18, 200), rng.normal(0.65, 0.18, 200)]
metrics = roc_metrics(y_true, y_score)
print(f"AUC: {metrics['auc']:.3f}")
```

```python
plt.plot(metrics["fpr"], metrics["tpr"], label=f"ROC (AUC={metrics['auc']:.3f})")
plt.plot([0, 1], [0, 1], "--", color="gray", label="Hasard")
plt.xlabel("FPR")
plt.ylabel("TPR")
plt.title("Courbe ROC — scores jouets")
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```

```python

import numpy as np
from scipy.stats import norm

def kgw_score(K, T, gamma=0.25):
    if T <= 0 or not (0 < gamma < 1) or not (0 <= K <= T):
        raise ValueError("Paramètres invalides")

    z = (K - gamma * T) / np.sqrt(T * gamma * (1 - gamma))
    p_value = norm.sf(z)  # Test unilatéral

    return {"z": z, "p_value": p_value}

print(kgw_score(K=80, T=200, gamma=0.25))
```
