# TextSeal ou détecteur jouet

Comparaison de deux scores synthétiques pour illustrer l'usage des métriques. Ce notebook ne reproduit pas TextSeal.

```python
from pathlib import Path
import sys

ROOT = next(parent for parent in [Path.cwd(), *Path.cwd().parents] if (parent / "src").is_dir())
sys.path.insert(0, str(ROOT))

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from src.metrics import roc_metrics
```

```python
rng = np.random.default_rng(7)
y_true = np.tile([0, 1], 150)
baseline = y_true + rng.normal(0, 0.9, y_true.size)
toy_detector = y_true + rng.normal(0, 0.55, y_true.size)

comparison = pd.DataFrame({
    "detector": ["baseline", "toy_detector"],
    "auc": [roc_metrics(y_true, score)["auc"] for score in (baseline, toy_detector)],
})
comparison
```

```python
comparison.plot.bar(x="detector", y="auc", ylim=(0.5, 1.0), legend=False, title="AUC — données synthétiques")
plt.ylabel("AUC")
plt.tight_layout()
plt.show()
```
