# Mock challenge

Exemple reproductible de bout en bout : génération de données factices, calcul de métriques, puis création d'un CSV de prédictions dans `outputs/`.

```python
from pathlib import Path
import sys

ROOT = next(parent for parent in [Path.cwd(), *Path.cwd().parents] if (parent / "src").is_dir())
sys.path.insert(0, str(ROOT))

import numpy as np
from src.metrics import confusion_rates, roc_metrics
from src.submission import make_submission, write_submission
```

```python
rng = np.random.default_rng(21)
y_true = np.tile([0, 1], 100)
scores = np.clip(y_true * 0.55 + rng.normal(0.25, 0.22, y_true.size), 0, 1)
print("AUC:", round(roc_metrics(y_true, scores)["auc"], 3))
print("Seuil 0.5:", confusion_rates(y_true, (scores >= 0.5).astype(int)))
```

```python
test_ids = [f"sample_{index:03d}" for index in range(10)]
test_scores = np.round(rng.uniform(0, 1, len(test_ids)), 5)
submission = make_submission(test_ids, test_scores)
output_path = write_submission(submission, ROOT / "outputs" / "mock_predictions.csv")
print(f"Fichier écrit : {output_path.relative_to(ROOT)}")
submission.head()
```
